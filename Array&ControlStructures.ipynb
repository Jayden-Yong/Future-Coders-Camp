## **Section 1: Lists**
So far, one variable holds one value:
```python
score = 85
```
But what if you have **30 students**? You don't want 30 variables.
A **list** lets you store many values under one name — in order, with an index starting at `0`.
```
 scores = [ 78,  85,  60,  92,  74 ]
 index:     0    1    2    3    4
```

#### Basic List Methods
| Method      | Description |
| :---        | :---        |
| `.append()` | Adds item to the end of the list |
| `.extend()` | Add elements of a list to the end of the current list |
| `.remove()` | Removes an item from the list |
| `.clear()`  | Removes all items from the list |
| `.index()`  | Returns the index of an item in the list |

#### Example 1 - Accessing items
Run the code below.

````python
scores = [78, 85, 60, 92, 74]

# Accessing items by index
print(scores[0])   
print(scores[2])   
print(scores[-1])  
````

> Notice that index `-1` returns the last item in the list.

#### Example 2 - Getting the length of a list
Run the code below.

````python
students = ["Aiman", "Nurul", "Wei Jie", "Priya"]
print(students[1])
print(len(students))
````

> Notice that `len(students)` returns the total item count in the list

#### Example 3 - Modifying a list
Run the code below.

````python
scores[2] = 75
scores.append(88)
scores.remove(92)

print(scores)
print("Length now:", len(scores))
````

## **Section 2: Control Structures**

### **For Loops**
A `for` loop goes through a list **one item at a time** and runs the same block of code for each one.  
The loop will stop at after a **specific** amount of loops.

#### Example 1 - Iterating over a list
Run the code below.

````python
students = ["Aiman", "Nurul", "Wei Jie", "Priya"]

for student in students:
    print("Hello,", student)
````

> Notice that the loop stops when all names in the list are displayed.

#### Example 2 - Iterating over a fixed range
Use the `range()` method to generate a sequence if numbers.  
Run the code below.

````python
for i in range(5):
    print("Round", i + 1)
````

> Notice that we print `i + 1`, that's because range(5) starts from `0` too.

#### Example 3
Run the code below.

````python
scores = [78, 85, 75, 88, 74]
total  = 0

for score in scores:
    total += score

average = total / len(scores)
print(f"Total: {total}")
print(f"Average: {round(average, 1)}")
````

> Notice that we print using `print(f"")`, these are called **f-strings**, used to format outputs and easily insert variable values into the output.

### **While Loops**
A `while` loop keeps running **as long as a condition is True**  
Useful when you don't know in advance how many times to repeat.

#### Example 1
Run the code below.

````python
countdown = 5

while countdown > 0:
    print(countdown)
    countdown -= 1

print("Go!")
````

> The loop will keep going as long as `countdown > 0` is `True`

#### Example 2
Run the code below.

````python
budget = 20.00
item_price = 3.50
items_bought = 0

while budget >= item_price:
    budget -= item_price
    items_bought += 1

print(f"Items bought: {items_bought}")
print(f"Change left: RM{round(budget, 2)}")
````

> **Infinite Loops**: This happens when the condition of the loop is **always `True`** and the loop will never stop.

#### When to use `for` or `while`?

| Use `for` when...                        | Use `while` when...                        |
|:---                                      |:---                                        |
| You know how many times to repeat        | You repeat until something changes         |
| Looping through a list                   | Waiting for user input or a condition      |
| `for item in list` / `for i in range(n)` | `while balance > 0` / `while not found`    |

````python
#Determining the gender based on the last 4 digits of your identity card
id_numbers = ['0123', '0406', '5009', '1475', '5422', '5446']

for last in id_numbers:
    last_digit = int(last[-1])
    if (last_digit % 2 == 0):
        print("Female")
    else: 
        print("Male")
````
