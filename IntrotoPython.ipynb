### Section 2a: Arithmetic Logics - Basics

### Section 2b: Arithmetic Logics - Comparison Operators
Comparison operators returns **booleans**, `True` or `False`.
| Operator | Meaning                  | Example    | Result  |
|----------|--------------------------|------------|---------|
| `==`     | Equal to                 | `5 == 5`   | `True`  |
| `!=`     | Not equal to             | `5 != 3`   | `True`  |
| `>`      | Greater than             | `7 > 10`   | `False` |
| `<`      | Less than                | `3 < 8`    | `True`  |
| `>=`     | Greater than or equal to | `6 >= 6`   | `True`  |
| `<=`     | Less than or equal to    | `9 <= 4`   | `False` |

##### Try running the cell below.

```python
price     = 12.50
money     = 10.00

can_afford = money >= price
print("Can I afford it?", can_afford)
```

<h3>Section 3: Boolean Logic</h3>

<p>Boolean logic is an integral part computer because computers rely on transistors that turn on and off to do arithmetic operations and logical operations. Boolean logic is used in programming to enable decisions to be made and is the basis to algorithms and control structures.</p>

<p>These Boolean operators are used for logical operation which returns a Boolean expression (<b>True</b> or <b>False</b>), example of which can be seen below: </p>

```python
#Sample of logical operations and it's outputs

value = 7
tax = 3.66
drink = "coca-cola"
state = True

#Sample of the AND operator
print( (value == 8) and (tax == 3.66) )
print( (drink == "coca-cola") and (state) )

#Sample of the OR operator
print( (value == 8) or (tax == 3.73) )
print( (drink == "pepsi") or (tax == 3.66) )
print( (state == True) or (value == 7) )

#Sample of the NOT operator 
print( not (value == 7) )
print( not ((drink == "pepsi") and (state)))
print( not ((drink == "pepsi") or (tax == 3.66)))
```

Boolean logic could also be chained and combined to become one long statement and yield one Boolean result

For example, the statement below is using to represent a logical operation:

`The value of x must be equal to 7 and the state of the machine must be set to be true, or, the value of y must be larger than 10 and the the volume of the tank must not equal to 70`

We can represent this long statement into a simple statement in Python code. Try to guess the final value of the Boolean statements below:

```python
#Sample statement above
x = 7
state_machine = True
y = 9
volume = 78
sample = ((x == 7) and (state_machine == True)) or ((y > 10) and (volume != 70))
print(sample)

#Statement No. 1
score = 50
name = "Alex"
is_logged_in = False

# A.
print( (score >= 50) and (name == "Alex") ) 

# B.
print( (score < 40) or (is_logged_in == True) )

# C.
print( not (name == "Jordan") )

#Statement No. 2
age = 16
has_permit = True
has_car = False

# D. 
print( (age >= 16) and (has_permit) and (has_car) )

# E.
print( (has_car == True) or (age > 10) )

# F.
print( (age == 16) and ((has_permit) or (has_car)) )

#Statement No. 3
temp = 32
weather = "Sunny"
is_weekend = True

# G.
print( not (temp > 40) )

# H.
print( not ((weather == "Rainy") or (is_weekend == False)) )

# I. (The Brain Tickler)
print( (not (temp == 0)) and (weather == "Sunny") )
```
