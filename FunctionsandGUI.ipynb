### **Section 1: Functions in Python**

Functions in Python are used to replace commonly used sections of a code or represent a specific set of operations that needed to be done.
Functions are also called procedures when it does not return any values.
A function is divided into multiple components:
- **Function Name**: is used to call the function within the code
- **Parameter**: the value that needs to passed to be used inside of the function
- **Return Value**: the value that is returned from the function

Here is the sample syntax for declaring a function in Python:

```python
    def function_name(params):
        #insert sample code here

        return values
```

Functions are useful since you don't have to rewrite the code all over again when you need to reuse the same section of code all over again.

**Example:**

Below is an example of a function in Python:

````python
#Sample Example: Converting between Fahrenheit and Celcius

def fahrenheit_celcius(temp, mode):
    if (mode.lower() == "celcius"):
        celcius = (temp - 32) * (5 / 9)
        return celcius
    elif (mode.lower() == "fahrenheit"):
        fahrenheit = (temp * (9/5)) + 32
        return fahrenheit
    else:
        return -1

temp = float(input("what is the current temperature: "))
mode = str(input("what is the mode you want: "))
print(f"The current temperature in {mode} is {fahrenheit_celcius(temp, mode):.2f}")
````

You can also pass arrays as a parameter or return value in a given function.

This is useful when you need to do an operation for a whole array instead of just variables.

**Example:**

The sample below is the same function as above but it passes an array instead.

````python
#Sample Example: Converting between Fahrenheit and Celcius using arrays

def convert_temp(temps, mode):
    converted = []
    if (mode.lower() == "celcius"):
        for i in range(len(temps)):
            temps[i] =  (temps[i] - 32) * (5 / 9)
            converted.append(temps[i])
    elif (mode.lower() == "fahrenheit"):
        for i in range(len(temps)):
            temps[i] = (temps[i] * (9/5)) + 32
            converted.append(temps[i])
    else:
        converted = []
    return converted


fahrenheit_temps = [32, 41, 50, 59, 68, 77, 86, 95, 104, 113]
print("Temp in Fahrenheit: ", fahrenheit_temps)
conv_fahrenheit = convert_temp(fahrenheit_temps, "celcius")
print("Converted Values: ", conv_fahrenheit)

celsius_temps = [0, 5, 10, 15, 20, 25, 30, 35, 40, 45]
print("Temp in Celcius: ", celsius_temps)
conv_celsius = convert_temp(celsius_temps, "fahrenheit")
print("Converted Temps: ", conv_celsius)
````

Functions are also called procedures when they do not pass any values using the `return` syntax.
Procedures are useful when you need to repeat specific lines code without rewriting the code.

**Example:**

The example below prints out the word "Congratulations!" based on the parameters set by the user.

````python
#Sample procedure

def print_congrats(num):
    for i in range(num):
        print("Congratulations!")

print_congrats(5)
print("===================")
print_congrats(5)
````

Another example of  a procedure without a user-defined parameter that prints the word "Congratulations!" seven times every time is is called.

````python
#Sample procedure

def print_congratulations():
    for i in range(7):
        print("Congratulations!")


print_congratulations()
print("===================")
print_congratulations()
````

### **Textboxes**

Textboxes are used for getting input from the user. The syntax are as follows:

> ```var_name = widgets.Text(description = 'text', placeholder = 'sample input') ```

There is also a variation for textboxes that recieve numerical inputs from the user. The syntax as follows:

> ```var_name = widgets.IntText(description = 'text', placeholder = 'sample input') ```

Then use the `display()` syntax to make the textbox appear.

You can use the variables that is set to the textbox as user-inputs for a function to be used in your program. The syntax is:

> ```name = var_name.value ```

**Example:**

The example below shows two textboxes, one for regular strings, one for number inputs

````python
#Textbox
textbox_sample = widgets.Text(description = "Sample Textbox", placeholder = "Sample Input")
display(textbox_sample)

#Integer Textbox
intbox_sample = widgets.IntText(description = "Sample IntTextbox", placeholder = "Sample Number")
display(intbox_sample)
````

**Example:**

The example below uses the input of the textbox in a function

````python
#Sample function
def print_name(name):
    print(f"Hello! My name is {name}. Nice to meet you!")

#Textbox input
textbox_name = widgets.Text(description = "Name", placeholder = "Please insert your name!")
name_button = widgets.Button(description = "click me!", button_style = "success")
display(textbox_name)
display(name_button)
name_button.on_click(lambda b: print_name(textbox_name.value))
````

### **Buttons**

Buttons are used for actions inside a GUI. The syntax to declare a button are as follows:

> ```var_name = widgets.Button(description = 'text', button_style = 'success/info/warning/danger') ```

Buttons must be used to call functions or procedures using the syntax below:

> ```var_name.on_click(lambda b: func_name) ```

Then use the `display()` syntax to display the button.

**Example:**

The code below displays all types of buttons: Success, Info, Warning and Danger

````python
#Success Button
sample_success = widgets.Button(description = "Success", button_style = "success")

#Info Button
sample_info = widgets.Button(description = "Info", button_style = "info")

#Warning Button
sample_warning = widgets.Button(description = "Warning", button_style = "warning")

#Danger Button
sample_danger = widgets.Button(description = "Danger", button_style = "danger")

#Displaying all buttons
display(sample_success)
display(sample_info)
display(sample_warning)
display(sample_danger)
````

**Example:**

The example below shows a button with a call to action. The button display Hello five times after it is clicked.

````python
#Declaring an output box and button
hello_output = widgets.Output()
hello_button = widgets.Button(description = "click me!", button_style = "info")

#Declaring a sample function
def print_hello():
    with hello_output:
        for i in range(5):
            print("Hello!")

#Displaying all items
hello_button.on_click(lambda b: print_hello())
display(hello_button)
display(hello_output)
````

## **Tables**

There is no defined function for a table in `ipywidgets`. Therefore, we need to use a little bit of HTML and the Output Box function to make tables in the GUI.

The syntax for a table are as follows:

> ```widgets.HTML("""<table>........</table>""")```

The syntax list for a table in HTML are as follows:
- `<table></table>` – Declares a table structure (you can add styles here as well)
- `<tr></tr>` – Declares a table row
- `<th></th>` – Declares a table header (bold by default)
- `<td></td>` – Declares table data for a given row

**Example:**

The example below is a pre-defined table:

````python
#Declaring the output box and the table

table_output = widgets.Output()
sample_table = widgets.HTML("""
                    <table border = '1'>
                    <tr>
                        <th> Header 1 </th>
                        <th> Header 2 </th>
                    </tr>
                    <tr>
                        <td> Data 1 </td>
                        <td> Data 2 </td>
                    </tr>
                    </table>""")

#Displaying the table
with table_output:
    display(sample_table)
display(table_output)
````

**Example:**

The example below prints the table using a function to append the HTML string

````python
#Declaring the data array
sample_data = [ ["Afiq", "21"],
                ["Mary", "22"],
                ["Syakirin", "23"] ]

#Declaring the string for HTML and the append function
str_html = "<table border = '1'>"

def append_str(str_html, data):
    for item in data:
        str_html += f"<tr><td>{item[0]}</td><td>{item[1]}</td></tr>"
    str_html += "</table>"
    return str_html

#Declare the sample textbox and displaying the
append_output = widgets.Output()
appended = append_str(str_html, sample_data)
append_table = widgets.HTML(appended)

with append_output:
    display(append_table)
display(append_output)
````

### **Layouts and Styling**

There are two main layouts that can be used in `ipywidgets`:
- `VBox()` displays widgets in a vertical layout
- `HBox()` displays widgets in a horizontal layout

The syntax for both is:
- ```var_name = widgets.HBox([list_of_widgets])```
- ```var_name = widgets.VBox([list_of_widgets])```

Layouts make your GUI more clean and structured by providing a box for the widgets to fit in.

**Example:**

The example below uses the same widgets but two different layouts

````python
#Declaring the widgets

lay_button = widgets.Button(description = "button")
lay_text = widgets.Text(desctiption = "textbox")
lay_output = widgets.Output()

with lay_output:
    print("output text")

#Declaring the layouts
vertical = widgets.HBox([lay_button, lay_text, lay_output])
horizontal = widgets.VBox([lay_button, lay_text, lay_output])
display(vertical)
display(horizontal)
````

## **Sample Program using ipywidgets GUI**

The sample program below is a sample program that converts an array of temperatures between Celcius and Fahrenheit or vice-versa.

The functionalities of the program are below:
- Recieve user inputs and append an array
- Converts the array into the desired temperature scale
- Clears the array with a button
- Displays the results in a table

````python
import ipywidgets as widgets
from IPython.display import display, clear_output

# 1. Arrays for data storage
temperatures = []
converted_temps = []

# 2. Widgets
add_button = widgets.Button(description="Add", button_style='info')
display_button = widgets.Button(description="Convert", button_style='success')
clear_button = widgets.Button(description="Clear", button_style="warning")

add_textbox = widgets.IntText(description="Insert Temperature", style={'description_width': 'initial'})
temp_dropdown = widgets.Dropdown(
    options=["Celsius", "Fahrenheit"],
    description="Target Scale",
    style={'description_width': 'initial'}
)

temp_output = widgets.Output()

# 3. Logic Functions
def update_display():
    with temp_output:
        clear_output(wait=True)
        str_html = "<table border='1' style='width:100%'><tr><th>Original</th><th>Converted</th></tr>"
        for orig, conv in zip(temperatures, converted_temps):
            str_html += f"<tr><td>{orig}</td><td>{conv:.2f}</td></tr>"
        str_html += "</table>"
        display(widgets.HTML(str_html))

def handle_submit(sender):
    if add_textbox.value != 0:
        temperatures.append(add_textbox.value)
        add_textbox.value = 0 # Clear the box
        update_display()

def handle_convert(sender):
    converted_temps.clear() # Clear old conversions
    mode = temp_dropdown.value

    for val in temperatures:
        if mode == "Celsius":
            res = (val - 32) * (5 / 9)
        else:
            res = (val * (9/5)) + 32
        converted_temps.append(res)

    update_display()

def handle_clear(sender):
    temperatures.clear()
    converted_temps.clear()
    update_display()

# 4. Linking Functionalities
add_button.on_click(handle_submit)
display_button.on_click(handle_convert)
clear_button.on_click(handle_clear)

# 5. Display Layout
layout = widgets.VBox([
    widgets.HBox([add_textbox]),
    widgets.HBox([temp_dropdown]),
    widgets.HBox([add_button, display_button, clear_button]),
    temp_output
])

display(layout)
update_display() # Initial empty table
````
